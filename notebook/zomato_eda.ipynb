```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# better visuals
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("T:\\2. PROJECTS\\5. Zomato Project\\Data\\raw\\zomato_restaurants_in_India.csv")
df.head()
```

```python
df.shape
df.columns
df.info()
```

```python
df.dtypes
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Drop useless columns
df = df.drop(columns=['zipcode', 'timings', 'address'])

# Handle missing cuisines
df = df.dropna(subset=['cuisines'])

# Fix opentable
df['opentable_support'] = df['opentable_support'].fillna(0).astype(int)
```

```python
df.isnull().sum().sort_values(ascending=False)
df.dtypes
```

```python
df.head(1)
```

```python
summary = pd.DataFrame({
    'dtype': df.dtypes,
    'nulls': df.isnull().sum(),
    'null_pct': (df.isnull().sum() / len(df)) * 100,
    'unique': df.nunique()
}).sort_values(by='null_pct', ascending=False)

summary
```

```python
df['weighted_rating'] = df['aggregate_rating'] * df['votes']

df['weighted_rating']
```

```python
df['value_score'] = df['aggregate_rating'] / df['average_cost_for_two']
```

```python
df['popularity'] = pd.qcut(df['votes'], q=3, labels=['Low', 'Medium', 'High'])
df['popularity']
```

```python
df['popularity'].value_counts()
```

```python
pd.qcut(df['votes'], q=3, labels=['Low','Medium','High'], duplicates='drop')
```

```python
df.groupby('popularity')['aggregate_rating'].mean().round(2)
```
